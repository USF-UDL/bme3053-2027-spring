## 1. Python and environment
This checks the Python version and that the notebook is using the course environment (the `.venv` folder created by `uv sync`).

```python
import platform
import sys

print("Python version:", platform.python_version())
print("Python location:", sys.executable)

problems = []
if ".venv" not in sys.executable:
    problems.append("Not using the course .venv environment (re-select the kernel)")

if problems:
    print("PROBLEM:", "; ".join(problems))
else:
    print("OK: Python and environment look right")
```

```python
import importlib

required = ["numpy", "scipy", "matplotlib", "pandas"]
missing = []

for name in required:
    try:
        module = importlib.import_module(name)
        print(f"  ok       {name} {getattr(module, '__version__', '')}")
    except ImportError:
        missing.append(name)
        print(f"  MISSING  {name}")

if missing:
    print("PROBLEM: missing", ", ".join(missing), "- run `uv sync` in the terminal, then restart the kernel")
else:
    print("OK: all packages installed")
```

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
print("mean =", x.mean())
print("squares =", x ** 2)

assert x.mean() == 3, "NumPy gave an unexpected result"
print("OK: NumPy works")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import cbook

with cbook.get_sample_data("s1045.ima.gz") as f: # type: ignore
    mri = np.frombuffer(f.read(), np.uint16).reshape((256, 256))

plt.imshow(mri, cmap="gray")
plt.title("Brain MRI slice")
plt.axis("off")
plt.show()
```

```python
import importlib
import sys

results = {}
results["Python >= 3.10"] = sys.version_info >= (3, 10)
results["Course .venv in use"] = ".venv" in sys.executable
for name in ["numpy", "scipy", "matplotlib", "pandas"]:
    try:
        importlib.import_module(name)
        results[f"import {name}"] = True
    except ImportError:
        results[f"import {name}"] = False

for check, passed in results.items():
    print(("  pass  " if passed else "  FAIL  ") + check)

print()
if all(results.values()):
    print(f"SETUP COMPLETE for {student_name}")
else:
    print("SETUP INCOMPLETE - see troubleshooting below")
```
